```python
import numpy
import pandas

import matplotlib.pyplot as plt
plt.style.use('../mystyle.mplstyle')

import sbruceana

PATH_TO_SBRUCE = "/Users/triozzi/Analysis/numine/sbruceana/data/cc1e0pi0p/"
```

#### Before NG2 HIP tag

```python
FILE_CV = "prefinal/CNAF_CV_1e0p0pi_NuMI_NoSysts.root"
FILE_OFFBEAM = "prefinal/CNAF_OffBeam_1e0p0pi_NuMI_NoSysts.root"
FILE_DATA = "prefinal/CNAF_Data_1e0p0pi_NuMI_NoSysts.root"
```

```python
# MC
df = sbruceana.io.convert_tree_to_df(
  f"{PATH_TO_SBRUCE}{FILE_CV}",
  "events/selectedNu"  
)
pot = sbruceana.utils.get_POT(f"{PATH_TO_SBRUCE}{FILE_CV}")

df_cos = sbruceana.io.convert_tree_to_df(
  f"{PATH_TO_SBRUCE}{FILE_CV}",
  "events/selectedCos"  
)
pot_cos = sbruceana.utils.get_POT(f"{PATH_TO_SBRUCE}{FILE_CV}")

df_cos['cosmic'] = 1
df = pandas.concat(
  (df, df_cos)
)

# data
df_data = sbruceana.io.convert_tree_to_df(
  f"{PATH_TO_SBRUCE}{FILE_DATA}",
  "events/selectedData"  
)
pot_data = sbruceana.utils.get_POT(f"{PATH_TO_SBRUCE}{FILE_DATA}")
time_data = sbruceana.utils.get_livetime_data(f"{PATH_TO_SBRUCE}{FILE_DATA}")

# offbeam
df_offbeam = sbruceana.io.convert_tree_to_df(
  f"{PATH_TO_SBRUCE}{FILE_OFFBEAM}",
  "offbeam/selectedOffbeam"  
)
time_offbeam = sbruceana.utils.get_livetime_offbeam(f"{PATH_TO_SBRUCE}{FILE_OFFBEAM}")
```

```python
fig, ax = plt.subplots(figsize=(4.25, 3.5), layout='constrained')

var = "collE"
width = 0.2; bins = numpy.arange(0.2, 2+width, width)

IS_AREA_NORMALIZED = True
ax = sbruceana.plotting.plot_by_category_with_offbeam(ax, df, sbruceana.config.CC1E0PI0P_CATEGORIES, bins, var, df_offbeam, offbeam_scale=time_data/time_offbeam, yscale=pot_data/pot, area_normalized=IS_AREA_NORMALIZED, band=True, clip=False)
ax = sbruceana.plotting.plot_data(ax, df_data, bins, var, IS_AREA_NORMALIZED)

# gfx
tag_count = '#'
if IS_AREA_NORMALIZED:
  tag_count = 'a.n.'
ax.set(
  title = f'NuMI MC: {pot:.1e} POT\nRun2 10% NuMI data: {pot_data:.1e} POT',
  xlabel = 'leading shower $E$ [GeV]',
  ylabel = f'slices [{tag_count}]\n/ {width} GeV',
  xlim   = (bins[0], bins[-1]),
)
leg = ax.legend(fontsize=9.5); leg.get_title().set_fontsize(11)

plt.show()
fig.savefig(f"plots/prefinal/prefinal_{var}.pdf", dpi=300)
```

```python
fig, ax = plt.subplots(figsize=(4.25, 3.5), layout='constrained')

var = "nngshr"
width = 1; bins = numpy.arange(1, 8+width, width)

IS_AREA_NORMALIZED = True
ax = sbruceana.plotting.plot_by_category_with_offbeam(ax, df, sbruceana.config.CC1E0PI0P_CATEGORIES, bins, var, df_offbeam, offbeam_scale=time_data/time_offbeam, yscale=pot_data/pot, area_normalized=IS_AREA_NORMALIZED, band=True, clip=False)
ax = sbruceana.plotting.plot_data(ax, df_data, bins, var, IS_AREA_NORMALIZED)

ax = sbruceana.plotting.place_cut(ax, 2, True)

# gfx
tag_count = '#'
if IS_AREA_NORMALIZED:
  tag_count = 'a.n.'
ax.set(
  title = f'NuMI MC: {pot:.1e} POT\nRun2 10% NuMI data: {pot_data:.1e} POT',
  xlabel = 'NG2-tagged showers [#]',
  ylabel = f'slices [{tag_count}]\n/ {width}',
  xlim   = (bins[0], bins[-1]),
)
leg = ax.legend(fontsize=9.5); leg.get_title().set_fontsize(11)

plt.show()
fig.savefig(f"plots/prefinal/prefinal_{var}.pdf", dpi=300)
```

```python
fig, ax = plt.subplots(figsize=(4.25, 3.5), layout='constrained')

var = "maxhipvtxhits"
width = 2; bins = numpy.arange(0, 50+width, width)

IS_AREA_NORMALIZED = True
ax = sbruceana.plotting.plot_by_category_with_offbeam(ax, df, sbruceana.config.CC1E0PI0P_CATEGORIES, bins, var, df_offbeam, offbeam_scale=time_data/time_offbeam, yscale=pot_data/pot, area_normalized=IS_AREA_NORMALIZED, band=True, clip=False)
ax = sbruceana.plotting.plot_data(ax, df_data, bins, var, IS_AREA_NORMALIZED)

# ax = sbruceana.plotting.place_cut(ax, 2, True)

ax = sbruceana.plotting.place_cut(ax, 5, True)

# gfx
tag_count = '#'
if IS_AREA_NORMALIZED:
  tag_count = 'a.n.'
ax.set(
  title = f'NuMI MC: {pot:.1e} POT\nRun2 10% NuMI data: {pot_data:.1e} POT',
  xlabel = 'NG2 HIP hits at vtx [#]',
  ylabel = f'slices [{tag_count}]\n/ {width}',
  xlim   = (bins[0], bins[-1]),
)
leg = ax.legend(fontsize=9.5); leg.get_title().set_fontsize(11)

ax.set_yscale('log')

plt.show()
fig.savefig(f"plots/prefinal/prefinal_{var}.pdf", dpi=300)
```
