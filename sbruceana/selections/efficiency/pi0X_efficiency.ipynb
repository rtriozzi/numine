```python
import numpy
import pandas

import matplotlib
import matplotlib.pyplot as plt
plt.style.use('../../mystyle.mplstyle')

import sbruceana

PATH_TO_DATA = '/Users/triozzi/Analysis/numine/sbruceana/data/'
```

```python
# import data
df = pandas.read_csv(
  f'{PATH_TO_DATA}pi0X/Pi0X_Efficiency.txt',
  names = ['trueE', 'pi0P', 'mincosgg', 'cospi0z', 'cospi0numi'] + [f'cut{i}' for i in range(12)],
  delimiter = '\t',
  index_col = False
)

CUT_LABELS = [
    'Presel.', 'FM', '2 $\\gamma$', 
    '$E_{\\gamma_1}$ >25 MeV', 'gap$_{\\gamma_1}$ >3 cm', '$\\theta_{\\nu\\gamma_1}$ >0.9', 'contained',
    '$E_{\\gamma_2}$ >25 MeV', 'gap$_{\\gamma_2}$ >3 cm', '$\\theta_{\\nu\\gamma_2}$ >0.9', 'contained',
    '$M_{\\pi^0}$ >0, <700 MeV'
]

CLEAN_CUTS = [1, 2, 6, 10]
CLEAN_CUT_LABELS = ['Pre-selection', '2 $\\gamma$', '$\\gamma_1$ quality', '$\\gamma_2$ quality']
```

```python
fig, ax = plt.subplots(figsize=(4.5, 4), layout='constrained')

var = 'trueE'
binning = numpy.arange(0, 5+0.2, 0.2)
ax = sbruceana.efficiency.plot_efficiency(ax, df, var, binning, range(12), CUT_LABELS[:])

ax.legend(fontsize=9, ncol=2, loc=(0., 1.15),
              framealpha=0.9, edgecolor='none')
```
