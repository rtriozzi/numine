##### "Standard" inclusive pi0 selection

This is not really a selection. I'm just looking for one pi0 across all interactions.
The aims here are: high statistics, good photons.

- v0: no cut on the photon-to-vertex direction; no cut on number of showers;
- v1: cut on the photon-to-vertex direction; no cut on number of showers;

Final version includes a cut on the number of NG2-tagged showers >25 MeV to be 2.

```python
df = sbruceana.io.convert_tree_to_df(
  f"{PATH_TO_SBRUCE}{FILE_CV}",
  "events/selectedNu"  
)
pot = sbruceana.utils.get_POT(f"{PATH_TO_SBRUCE}{FILE_CV}")

df_cos = sbruceana.io.convert_tree_to_df(
  f"{PATH_TO_SBRUCE}{FILE_CV}",
  "events/selectedCos"  
)
pot_cos = sbruceana.utils.get_POT(f"{PATH_TO_SBRUCE}{FILE_CV}")

df_cos['cosmic'] = 1
df = pandas.concat(
  (df, df_cos)
)

# offbeam
df_offbeam = sbruceana.io.convert_tree_to_df(
  f"{PATH_TO_SBRUCE}{FILE_OFFBEAM}",
  "offbeam/selectedOffbeam"  
)
time_offbeam = sbruceana.utils.get_livetime_offbeam(f"{PATH_TO_SBRUCE}{FILE_OFFBEAM}")

# data
df_data = sbruceana.io.convert_tree_to_df(
  f"{PATH_TO_SBRUCE}{FILE_DATA}",
  "events/selectedData"  
)
pot_data = sbruceana.utils.get_POT(f"{PATH_TO_SBRUCE}{FILE_DATA}")
time_data = sbruceana.utils.get_livetime_data(f"{PATH_TO_SBRUCE}{FILE_DATA}")
```

```python
def AddCorrectedPi0Mass(
    df,
    FACTOR = 1.2
):
  df["pi0mass_corr"] = \
    numpy.sqrt(2 * (1e3*df['leadcollE']*FACTOR) * (1e3*df['sublcollE']*FACTOR) * (1 - df['pi0open'])) # MeV
  return df
```

```python
IS_AREA_NORMALIZED = False
```

```python
fig, ax = plt.subplots(figsize=(4.25, 3.5), layout='constrained')

var = "simplepi0mass"

width = 15; bins = numpy.arange(15, 390+width, width)

NoEtasHere = (df["simplepi0mass"] < 390)
ax = sbruceana.plotting.plot_by_category_with_offbeam(ax, df[NoEtasHere], sbruceana.config.PI0X_CATEGORIES, bins, var, df_offbeam, offbeam_scale=time_data/time_offbeam, yscale=pot_data/pot, area_normalized=IS_AREA_NORMALIZED, band=True)

NoEtasHere = (df_data["simplepi0mass"] < 390)
ax = sbruceana.plotting.plot_data(ax, df_data[NoEtasHere], bins, var, IS_AREA_NORMALIZED, color='black', label='data')

ax.axvline(MASS_PI0, zorder=3, lw=0.75, c='black', label='134.9766 MeV')

if IS_AREA_NORMALIZED:
  FACTOR = 1.20
  df = AddCorrectedPi0Mass(df, FACTOR)
  ax.hist(df["pi0mass_corr"], bins=bins, density=True, histtype='step', color='black', linewidth=1.5, label=f'$E$$\\times${FACTOR}')

# gfx
ax.set(
  title = f'NuMI MC: {pot:.1e} POT\nRun2 10% NuMI data: {pot_data:.1e} POT',
  xlabel = '$M_{\\pi^0}$ [MeV]',
  ylabel = f'slices [#]\n/ {width} MeV',
  xlim   = (bins[0], bins[-1]),
)
# leg = ax.legend(fontsize=9.5, title='NuMI CV'); leg.get_title().set_fontsize(10)
leg = ax.legend(fontsize=9.5)

# fig.suptitle("All uncorrected", fontsize=20)

plt.show()
fig.savefig(f"plots/calorimetry_fixed_{var}.pdf", dpi=300)
fig.savefig(f"plots/calorimetry_fixed_{var}.png", dpi=300)
```

##### Photon energy calibration

- μBooNE reports a flat factor of 0.83, which would be reasonable for ICARUS;
- SBND reported a similar conclusion, plus a significant bias at higher photon energies, leading to funny splines.

SBND method: create bins of E_reco, augmenting them by using overlapping bins (e.g.: 0->0.1, 0.01>0.11, ...). In each bin, compute the average E_true/E_reco factor by fitting its distribution. Then, fit the resulting trend or use a spline.

```python
# create simple non-overlapping bins in true energy
edges_true = numpy.array([0.025, 0.1, 0.2, 0.3, 0.5, 1.])
edges      = list(zip(edges_true[:-1], edges_true[1:]))

# compute the (E_reco - E_true) / E_true resolution
resid       = numpy.where(X > 0, (Y - X) / X, numpy.nan)
bins_ratio  = numpy.arange(-1, 0.3+0.05, 0.05)
bin_centers = 0.5 * (bins_ratio[:-1] + bins_ratio[1:])

# plot the (E_reco-E_true)/E_true resolution in each true-energy bin
n_bins  = len(edges)
n_cols  = 5
n_rows  = int(numpy.ceil(n_bins / n_cols))

fig, axes = plt.subplots(
    n_rows, n_cols,
    figsize = (n_cols * 2.5, n_rows * 2.0),
    layout  = 'constrained',
)
axes_flat = axes.flat

mus, sigmas         = [], []
mus_err, sigmas_err = [], []
for ax, (lo, hi) in zip(axes_flat, edges):
    # find the resolution values whose E_true falls in this bin
    in_bin  = (X >= lo) & (X < hi)
    r_vals  = resid[in_bin]
    r_valid = r_vals[numpy.isfinite(r_vals)]

    # binned resolution in each E_true bin
    hist_vals, _ = numpy.histogram(r_valid, bins=bins_ratio)
    ax.hist(r_valid, bins=bins_ratio, fc='C0', histtype='stepfilled',
            label='$E_\\mathrm{true}$ bin\n'+f'[{lo:.2f}, {hi:.2f})\nn={len(r_valid)}')

    # fit: left-tailed CB
    mu_fit, sigma_fit         = numpy.nan, numpy.nan
    mu_err_fit, sigma_err_fit = numpy.nan, numpy.nan
    if len(r_valid) >= 10:
        try:
            cut    = (r_valid > -1) & (r_valid < 2)
            p0     = [hist_vals.max(), numpy.median(r_valid[cut]), numpy.std(r_valid[cut]), 1.5, 5] # A, mu, sigma, alpha, n
            bounds = ([0, -1, 1e-4, 0.05, 1.01], [numpy.inf, 2, 2, 10, 100])
            popt, pcov = scipy.optimize.curve_fit(
                sbruceana.fitting.crystal_ball_left, bin_centers, hist_vals,
                p0=p0, bounds=bounds, maxfev=10000,
            )
            perr = numpy.sqrt(numpy.diag(pcov))

            A_f, mu_f, sigma_f, alpha_f, n_f = popt
            A_e, mu_e, sigma_e, alpha_e, n_e = perr

            mu_fit, sigma_fit         = mu_f, sigma_f
            mu_err_fit, sigma_err_fit = mu_e, sigma_e

            x_fit = numpy.linspace(bins_ratio[0], bins_ratio[-1], 300)
            fit_label = f'$\\mu$={mu_f:.3f}\n$\\sigma$={sigma_f:.3f}'
            ax.plot(x_fit, sbruceana.fitting.crystal_ball_left(x_fit, *popt),
                    c='red', lw=1.25, ls='-', label=fit_label)
        except RuntimeError:
            print(f"Broken fit in bin [{lo:.2f}, {hi:.2f}).")
    mus.append(mu_fit)
    sigmas.append(sigma_fit)
    mus_err.append(mu_err_fit)
    sigmas_err.append(sigma_err_fit)

    # gfx
    ax.axvline(0, c='black', lw=0.75)
    ax.set_xlim(bins_ratio[0], bins_ratio[-1])
    ax.legend(loc='upper left', fontsize=7, handlelength=1, columnspacing=1)

# hide any unused axes
for ax in list(axes_flat)[n_bins:]:
    ax.set_visible(False)

# gfx
fig.suptitle('$(E_\\mathrm{reco}-E_\\mathrm{true})/E_\\mathrm{true}$ for $\\gamma$s in 1$\\pi^0$$X$', fontsize=30)

plt.show()
fig.savefig(f"plots/photon_calibration_resolution.pdf", dpi=300)
```

```python
MASK_MC = (df.simplepi0mass > 0) & (df.simplepi0mass < 250)
mc = numpy.array(df[MASK_MC].simplepi0mass.copy())

MASK_DATA = (df_data.simplepi0mass > 0) & (df_data.simplepi0mass < 250)
data = numpy.array(df_data[MASK_DATA].simplepi0mass.copy())
```

```python
# --- binning for the peak fit (matches the [0, 250] MeV mask above) ---
BINS = numpy.arange(0, 250 + 15, 15)
CTRS = 0.5 * (BINS[:-1] + BINS[1:])

# --- fit the full MC once (all 7 params) to fix the tail shape ---
def fit_full(sample):
    c  = numpy.histogram(sample, bins=BINS)[0].astype(float)
    p0 = [c.max(), numpy.median(sample), 25., 1.5, 5., 1.5, 5.]
    lo = [0,      80,  5,  0.3, 1.05, 0.3, 1.05]
    hi = [numpy.inf, 200, 80, 10,  50,   10,  50]
    with numpy.errstate(invalid='ignore'):
        popt, _ = scipy.optimize.curve_fit(
            sbruceana.fitting.crystal_ball_double, CTRS, c,
            p0=p0, bounds=(lo, hi), maxfev=20000)
    return popt

TAILS = fit_full(mc)[3:]   # (alpha_l, n_l, alpha_r, n_r) fixed from high-stat MC
print(f"Fixed tail params (from MC): alpha_l={TAILS[0]:.2f}, n_l={TAILS[1]:.2f}, "
      f"alpha_r={TAILS[2]:.2f}, n_r={TAILS[3]:.2f}")

def cb_fixed_tails(x, A, mu, sigma):
    return sbruceana.fitting.crystal_ball_double(x, A, mu, sigma, *TAILS)

def fit_mu(sample):
    """Stable 3-param fit (A, mu, sigma) with tails fixed. Returns mu or nan."""
    c = numpy.histogram(sample, bins=BINS)[0].astype(float)
    if c.sum() < 20:
        return numpy.nan
    p0 = [c.max(), CTRS[numpy.argmax(c)], 20.]
    try:
        with numpy.errstate(invalid='ignore'):
            popt, _ = scipy.optimize.curve_fit(
                cb_fixed_tails, CTRS, c, p0=p0,
                bounds=([0, 80, 5], [numpy.inf, 200, 80]), maxfev=20000)
        return popt[1]
    except Exception:
        return numpy.nan
```

```python
rng    = numpy.random.default_rng(42)
N      = len(data)        # data-sized draws
N_BOOT = 5000

mu_mc   = fit_mu(mc)      # MC peak position — the reference "truth"
mu_data = fit_mu(data)    # single observed data peak position (NOT bootstrapped)

# null distribution: fit mu on data-sized subsamples of the MC pool.
# This is "what would a 250-event measurement of the MC truth look like?"
mu_sub = numpy.array([
    fit_mu(rng.choice(mc, size=N, replace=False))
    for _ in range(N_BOOT)
])
mu_sub = mu_sub[numpy.isfinite(mu_sub)]

c       = mu_sub.mean()                 # null centre (≈ mu_mc; cross-check they agree)
dmu_obs = mu_data - mu_mc

# p-value = fraction of data-sized MC draws at least as far from the MC truth
p_two = numpy.mean(numpy.abs(mu_sub - c) >= numpy.abs(mu_data - c))
p_one = numpy.mean((mu_sub - c) >= (mu_data - c))   # data > MC expected direction

print(f"survived subsamples : {len(mu_sub)} / {N_BOOT}")
print(f"mu_MC (full)        : {mu_mc:.2f} MeV   null centre = {c:.2f} MeV   "
      f"null std = {mu_sub.std(ddof=1):.2f} MeV")
print(f"mu_data (observed)  : {mu_data:.2f} MeV")
print(f"shift               : {dmu_obs:.2f} MeV")
print(f"p (two-sided) = {p_two:.3f}    p (one-sided) = {p_one:.3f}")
print(f"=> data peak {'in tension with' if p_two < 0.05 else 'compatible with'} "
      f"MC at the 95% CL")
```
