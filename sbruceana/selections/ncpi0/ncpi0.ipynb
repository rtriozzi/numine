##### Selection studies -- v0

This had a muon veto (which was not working *that* well), the proton selection and "other particle" veto.
The photon selection was not super clean, so a lot of split nues (at very-forward angles) sneaked in.

```python
PATH_TO_SBRUCE = "/Users/triozzi/Analysis/numine/sbruceana/data/ncpi0/"

FILE_CV = "final/CNAF_CV_NCpi0_NuMI_NoSysts_v0.root"
```

```python
df = sbruceana.io.convert_tree_to_df(
  f"{PATH_TO_SBRUCE}{FILE_CV}",
  "events/selectedNu"  
)
pot = sbruceana.utils.get_POT(f"{PATH_TO_SBRUCE}{FILE_CV}")

df_cos = sbruceana.io.convert_tree_to_df(
  f"{PATH_TO_SBRUCE}{FILE_CV}",
  "events/selectedCos"  
)
pot_cos = sbruceana.utils.get_POT(f"{PATH_TO_SBRUCE}{FILE_CV}")

df_cos['cosmic'] = 1
df = pandas.concat(
  (df, df_cos)
)
```

##### Selection studies -- v1

No muon veto.
Cleaned photon selection:
- with energies between 25 MeV and 1 GeV;
- conversion gaps > 5 cm (not very forgiving, but still);
- reconstructed pi0 mass between 0 and 600 MeV.

Main background is given by other interactions with pi0s, for example muon neutrinos and other NC (where there are multiple pi0s, for example). 
A possible way out is to add thresholds to the pi0, to avoid low-momentum pi0 sneaking in (as they might not be really "visible"). 
Also, the "other particle" veto is unforgiving with split showers.
